# EDA + PCA on MNIST

Same idea as the dimension-reduction seminar and the eigenfaces project, just on digits instead of faces: look at what PCA keeps, how much variance the first few components explain, and whether unsupervised clustering on those components lines up with the true digit labels.

```python
import sys
sys.path.insert(0, "..")

import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

from src.data import flatten, load_mnist_arrays

X_train, y_train, X_test, y_test = load_mnist_arrays()
print(X_train.shape, y_train.shape)
```

```python
# a few example digits
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.flat):
    ax.imshow(X_train[i], cmap="gray")
    ax.set_title(str(y_train[i]))
    ax.axis("off")
plt.show()
```

```python
X_flat = flatten(X_train)
pca = PCA(n_components=2)
proj = pca.fit_transform(X_flat[:5000])

plt.figure(figsize=(7, 6))
scatter = plt.scatter(proj[:, 0], proj[:, 1], c=y_train[:5000], cmap="tab10", s=6, alpha=0.6)
plt.colorbar(scatter, label="digit")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("First 2 PCA components of MNIST pixels")
plt.show()
```

```python
pca_full = PCA().fit(X_flat[:5000])
plt.plot(np.cumsum(pca_full.explained_variance_ratio_))
plt.xlabel("n components")
plt.ylabel("cumulative explained variance")
plt.title("Scree plot \u2014 how many components to keep?")
plt.show()
```
